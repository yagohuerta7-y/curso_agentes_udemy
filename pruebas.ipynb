```python
def solution(roman : str) -> int:
    
    valores = {
        'I': 1,
        'V': 5,
        'X': 10,
        'L': 50,
        'C': 100,
        'D': 500,
        'M': 1000,
    }
    
    suma = 0
    
    for letra in set(roman):
        suma = suma + roman.count(letra)*valores[letra]
        
    return suma
```

```python
solution("v")
```
